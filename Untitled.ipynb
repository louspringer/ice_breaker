```python
# Start the scraping process
def start_scraping(search_url):
    # search_url = "https://www.linkedin.com/search/results/people/?keywords=YOUR_QUERY"  # Replace YOUR_QUERY with actual keywords
    driver.get(search_url)
    print("Navigated to the LinkedIn search page.")
    
    # Wait for the results to load
    WebDriverWait(driver, 10).until(
        EC.presence_of_element_located((By.XPATH, '//*[@data-view-name="search-entity-result-universal-template"]'))
    )
    
    # Begin scraping and pagination
    paginate_and_scrape()
    save_data()
    driver.quit()
```

```python
def scrape_page():
    results = []
    # Locate all search result containers
    results_list = driver.find_elements(By.XPATH, '//*[@data-view-name="search-entity-result-universal-template"]')

    for index, result in enumerate(results_list):
        try:
            # Extract the profile link and name
            profile_link_element = result.find_element(By.XPATH, './/a[contains(@href, "/in/")]')
            href = profile_link_element.get_attribute("href").split('?')[0]  # Clean URL
            name = profile_link_element.text.strip()

            # Extract the title
            try:
                title_element = result.find_element(By.XPATH, './/div[contains(@class, "entity-result__primary-subtitle")]')
                title = title_element.text.strip()
            except:
                title = None

            # Extract the location
            try:
                location_element = result.find_element(By.XPATH, './/div[contains(@class, "entity-result__secondary-subtitle")]')
                location = location_element.text.strip()
            except:
                location = None

            # Extract the mutual connections
            try:
                mutual_connections_element = result.find_element(By.XPATH, './/span[contains(@class, "entity-result__simple-insight-text")]')
                mutual_connections = mutual_connections_element.text.strip()
            except:
                mutual_connections = None

            # Append the data as a dictionary
            results.append({
                "Name": name,
                "Profile Link": href,
                "Title": title,
                "Location": location,
                "Mutual Connections": mutual_connections,
            })
        except Exception as e:
            print(f"Error processing result {index + 1}: {e}")

    return results
```

```python
def paginate_and_scrape():
    """Scrape data across multiple pages."""
    global df
    page = 1
    while True:
        print(f"Scraping page {page}...")
        page_data = scrape_page()
        df = pd.concat([df, pd.DataFrame(page_data)], ignore_index=True)
        print(f"Page {page} scraped. Total rows so far: {len(df)}")

        try:
            next_button = WebDriverWait(driver, 10).until(
                EC.presence_of_element_located((By.XPATH, '//button[contains(@aria-label, "Next") and contains(@class, "artdeco-pagination__button--next")]'))
            )
            driver.execute_script("arguments[0].scrollIntoView(true);", next_button)  # Ensure visibility
            ActionChains(driver).move_to_element(next_button).perform()

            if not next_button.is_enabled() or "disabled" in next_button.get_attribute("class"):
                print("Next button is disabled or not interactable. Stopping pagination.")
                break

            next_button.click()
            time.sleep(3)  # Add delay to wait for the next page to load
            page += 1
        except Exception as e:
            print(f"Pagination complete or issue encountered: {e}")
            break

def save_data():
    """Save the scraped data to a CSV."""
    file_path = 'linkedin_profiles_cleaned.csv'
    df.to_csv(file_path, index=False)
    print(f"Data saved to {file_path}")
```

```python
from selenium.webdriver.common.by import By
from selenium.webdriver.remote.webelement import WebElement

def extract_attributes(element: WebElement) -> dict:
    """
    Extract all attributes and their values for a given WebElement.
    """
    return element.parent.execute_script(
        '''
        let attrs = arguments[0].attributes;
        let result = {};
        for (let i = 0; i < attrs.length; i++) {
            result[attrs[i].name] = attrs[i].value;
        }
        return result;
        ''', 
        element
    )

# Expand individual results
expanded_results = [
    {
        'text': result.text,
        'attributes': extract_attributes(result)
    }
    for result in results
]

# Print the expanded results
for idx, result in enumerate(expanded_results, start=1):
    print(f"Result {idx}:")
    print(f"Text: {result['text']}")
    print("Attributes:")
    for attr, value in result['attributes'].items():
        print(f"  {attr}: {value}")
    print()
```

```python
from urllib.parse import urlsplit, urlunsplit

from selenium.webdriver.remote.webelement import WebElement
from typing import List

from urllib.parse import urlsplit, urlunsplit
from selenium.webdriver.remote.webelement import WebElement
from typing import List

from urllib.parse import urlsplit, urlunsplit
from selenium.webdriver.remote.webelement import WebElement
from typing import List

def strip_query_string(elements: List[WebElement]) -> List[str]:
    """
    Extracts 'href' attributes from Selenium WebElements, removes query strings
    and fragments, and returns a list of cleaned URLs.

    Args:
        elements (List[WebElement]): List of Selenium WebElement objects.
    
    Returns:
        List[str]: List of URLs with query strings and fragments removed.
    """
    return [
        urlunsplit((parsed_url.scheme, parsed_url.netloc, parsed_url.path, '', ''))
        for parsed_url in map(
            urlsplit, 
            (element.get_attribute("href") for element in elements if element.get_attribute("href"))
        )
    ]
```
